````python
**LÝ THUYẾT QUY TRÌNH — GIAI ĐOẠN 1**

**1. Xác định câu hỏi phân tích (Define the question)**

Xác định mục tiêu phân tích **trước khi tải và làm sạch dữ liệu**:

- Muốn tìm hiểu hoặc trả lời vấn đề gì?
- Kết quả phân tích phục vụ đối tượng nào?
- Đánh giá mức độ đạt mục tiêu bằng những chỉ số hoặc tiêu chí nào?

**2. Nạp dữ liệu (Data loading)**

- **CSV:** sử dụng `read_csv`.
- **Excel:** sử dụng `read_excel`.
- **Kaggle Hub:** tải bộ dữ liệu bằng `dataset_download`, sau đó đọc bằng hàm `read_*` phù hợp.
- **Nối dữ liệu:** sử dụng `concat` để nối các bảng theo hàng hoặc cột.
- **Ghép dữ liệu:** sử dụng `merge` để kết hợp các bảng theo khóa chung.

**3. Cơ chế dữ liệu thiếu (Missingness mechanisms)**

Cơ chế dữ liệu thiếu mô tả **nguyên nhân một giá trị không được quan sát hoặc ghi nhận**, được phân thành:

- **MCAR — Missing Completely At Random: Thiếu hoàn toàn ngẫu nhiên.**  
  Khả năng thiếu không phụ thuộc vào giá trị của các biến đã quan sát hoặc chưa quan sát.

- **MAR — Missing At Random: Thiếu ngẫu nhiên có điều kiện.**  
  Khả năng thiếu có thể phụ thuộc vào các biến đã quan sát; sau khi xét các biến đó, không còn phụ thuộc vào chính giá trị bị thiếu.

- **MNAR — Missing Not At Random: Thiếu không ngẫu nhiên.**  
  Khả năng thiếu vẫn phụ thuộc vào chính giá trị chưa được quan sát, ngay cả sau khi xét các biến đã quan sát.

**Xử lý giá trị thiếu theo loại biến**

| Trường hợp | Phương án xử lý cơ bản |
|---|---|
| Biến số có ít giá trị thiếu và không phải biến mục tiêu | Điền bằng **trung bình (mean)**; ưu tiên **trung vị (median)** nếu phân phối lệch |
| Biến văn bản hoặc biến phân loại | Điền bằng **giá trị xuất hiện nhiều nhất — mốt (mode)** |
| Thiếu biến mục tiêu | Dùng `dropna` để loại các dòng thiếu nhãn khỏi phần phân tích cần biến mục tiêu |
| Giá trị không hợp lệ, ví dụ `Age = −1` | Xác minh hoặc chuyển thành giá trị thiếu, sau đó lựa chọn cách xử lý phù hợp |
| Một dòng thiếu quá nhiều thông tin cần thiết | Cân nhắc loại dòng bằng `dropna` |
| Nhiều câu trả lời "Không muốn tiết lộ" (*Prefer not to say*), ví dụ cột `Cafe` | Có thể biểu diễn bằng nhóm `"Unknown"` — "Không rõ" |

Sau khi loại dòng, có thể dùng `reset_index(drop=True)` để **đánh lại chỉ mục**, không giữ chỉ mục cũ.

*Lưu ý: Đây là các phương án cơ bản, cần lựa chọn theo mục tiêu và cơ chế thiếu. `dropna` không trực tiếp loại giá trị −1 vì −1 chưa phải giá trị thiếu.*

**4. Chuẩn bị và biến đổi dữ liệu (Data Wrangling)**

Quy trình gồm:

**Khám phá → Tổ chức cấu trúc → Làm sạch → Bổ sung thông tin → Kiểm tra tính hợp lệ → Công bố dữ liệu**

| Giai đoạn | Nội dung |
|---|---|
| **Discovery — Khám phá** | Tìm hiểu nguồn, cấu trúc, ý nghĩa và chất lượng dữ liệu |
| **Structuring — Tổ chức cấu trúc** | Đưa dữ liệu về cấu trúc phù hợp để xử lý và phân tích |
| **Cleaning — Làm sạch** | Xử lý dữ liệu thiếu, sai, trùng lặp hoặc không nhất quán |
| **Enriching — Bổ sung thông tin** | Kết hợp nguồn dữ liệu khác hoặc tạo thêm biến có ý nghĩa |
| **Validating — Kiểm tra tính hợp lệ** | Kiểm tra kiểu dữ liệu, miền giá trị, tính nhất quán và các ràng buộc |
| **Publishing — Công bố dữ liệu** | Xuất hoặc cung cấp dữ liệu đã chuẩn bị để sử dụng tiếp |

**5. Phân tích khám phá dữ liệu (Exploratory Data Analysis — EDA)**

**EDA = Thống kê mô tả + Trực quan hóa dữ liệu.**

Mục tiêu là khám phá đặc điểm, xu hướng và mối liên hệ trong dữ liệu. EDA **chưa phải xây dựng mô hình học máy** và không đủ để khẳng định quan hệ nhân quả.

**Các bước thực hiện EDA**

*Tài liệu gốc ghi "6 bước" nhưng liệt kê 7 mục; có thể hiểu là 6 bước phân tích và 1 bước tổng kết.*

1. **Nạp và tìm hiểu dữ liệu:** kiểm tra `shape`, cấu trúc và ý nghĩa các cột.
2. **Kiểm tra dữ liệu thiếu và giá trị không hợp lệ:** xác định mức độ, đặc điểm và phương án xử lý.
3. **Kiểm tra giá trị ngoại lệ (outliers):** sử dụng **khoảng tứ phân vị (IQR)**; cân nhắc loại bỏ, giới hạn giá trị (*clipping*) hoặc xử lý theo phân vị (*winsorization*).
4. **Khảo sát phân phối:** sử dụng `describe`, biểu đồ phân phối tần suất (*histogram*) và biểu đồ đếm (*countplot*).
5. **Khảo sát mối quan hệ:** sử dụng biểu đồ nhiệt (*heatmap*), biểu đồ hộp (*boxplot*) hoặc biểu đồ violin; ưu tiên so sánh **tỷ lệ** thay vì chỉ số lượng khi quy mô các nhóm khác nhau.
6. **Tạo và biến đổi đặc trưng (feature engineering):** xử lý ngày giờ, tạo biến mới, chia khoảng (*binning*), mã hóa biến giả (*dummy encoding*) và chuẩn hóa thang đo (*scaling*).
7. **Tổng kết:** trình bày các phát hiện chính cùng những hạn chế và lưu ý khi diễn giải.

**6. Trực quan hóa dữ liệu (Visualization) — Nội dung tự học**

**Matplotlib — Kiểm soát chi tiết biểu đồ**

```python
import matplotlib.pyplot as plt
```

Matplotlib là thư viện phổ biến để tạo biểu đồ hai chiều trong Python.

- **Hai cách sử dụng API:**
  - Giao diện Pyplot: ví dụ `plt.plot`.
  - Giao diện hướng đối tượng: ví dụ `fig, ax = plt.subplots()`.
- **Các biểu đồ cơ bản:** đường (*line*), phân tán (*scatter*), cột (*bar*) và phân phối tần suất (*histogram*).
- **Bố cục:** sử dụng `subplots` để đặt nhiều vùng biểu đồ (*Axes*) trong cùng một hình (*Figure*).
- **Trường hợp sử dụng:** cần tùy chỉnh tiêu đề, giới hạn trục, bố cục và xuất hình bằng `savefig`.

**Seaborn — Trực quan hóa thống kê nhanh từ DataFrame**

```python
import seaborn as sns
```

Seaborn là thư viện trực quan hóa thống kê được xây dựng trên Matplotlib và sử dụng kết hợp với Matplotlib.

- **Dữ liệu đầu vào:** thường sử dụng các cột trong DataFrame; tham số `hue` phân biệt nhóm bằng màu sắc.
- **Các biểu đồ:** phân tán, đường, cột, đếm, histogram, hộp, violin, nhiệt và ma trận biểu đồ cặp (*pairplot*).
- **Biểu đồ phân ô (facets):** sử dụng `relplot` hoặc `catplot` với `col`, `row` để tạo các biểu đồ riêng theo nhóm.
- **Thiết lập giao diện:** sử dụng `sns.set_theme()`.
- **Trường hợp sử dụng:** thực hiện EDA nhanh; tiếp tục sử dụng `ax` hoặc `plt` để hoàn thiện và lưu biểu đồ.
````

**Chất Lượng Dữ Liệu & Cơ Chế Dữ Liệu Khuyết (Rubin's Missing Data Theory)**

**1. Phân loại cơ chế thiếu dữ liệu**

Được phát triển bởi Donald Rubin, việc nhận diện chính xác nguyên nhân thiếu dữ liệu giúp định hướng phương pháp xử lý:

- **MCAR (Missing Completely at Random):** Thiếu hoàn toàn ngẫu nhiên, không phụ thuộc vào bất kỳ biến quan sát hay biến ẩn nào (ví dụ: lỗi form ngẫu nhiên vài dòng).
- **MAR (Missing at Random):** Dữ liệu thiếu có liên quan đến các biến đã được quan sát khác trong tập dữ liệu (ví dụ: một nhóm độ tuổi cụ thể có xu hướng không điền câu hỏi thu nhập).
- **MNAR (Missing Not at Random):** Dữ liệu thiếu phụ thuộc chính vào giá trị bị ẩn giấu (ví dụ: người có thu nhập quá cao hoặc quá thấp chủ động từ chối khai báo). Đây là dạng khó xử lý nhất, dễ gây chệch lệch nghiêm trọng nếu điền khuyết sai cách.

**2. Chiến lược điền khuyết (Imputation) theo kiểu cột:**

- **Cột số (Numeric):** Nếu ít missing và không phải biến mục tiêu (target) -> dùng Mean (hoặc Median nếu phân phối lệch).
- **Cột văn bản / phân loại (Categorical):** Điền bằng Mode (giá trị xuất hiện nhiều nhất).
- **Trường hợp loại bỏ (dropna):** Áp dụng cho biến mục tiêu bị thiếu, giá trị không hợp lệ (ví dụ: tuổi = -1), quá nhiều hàng trống, hoặc dữ liệu thiếu thuộc dạng MCAR/MAR với tỷ lệ rất ít.
- **Trường hợp "từ chối trả lời" (ví dụ: khảo sát):** Điền nhãn chuyên biệt như "Unknown".

---

**Xử Lý Dữ Liệu: Outlier, Biến Đổi & Phân Phối**

**1. Xử lý giá trị ngoại lai (Outliers)**

- **Xác định:** Thường sử dụng phương pháp khoảng tứ phân vị IQR = Q3 - Q1, giới hạn ngoại lai nằm ngoài ngưỡng [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR].
- **Các hướng xử lý chính (3 hướng):**
  - Xóa bỏ bản ghi (drop).
  - Thay thế hoặc điền khuyết giá trị (impute).
  - Biến đổi hoặc giữ nguyên (transform / clip / winsorize).

**2. Hình dạng phân phối dữ liệu (Distribution Shapes & Skewness)**

Ảnh hưởng trực tiếp đến việc chọn thước đo đại diện (Mean hay Median) để điền khuyết:

- **Phân phối cân bằng (Symmetrical / Normal):** Đồ thị hình chuông đối xứng; Mean ≈ Median. Có thể dùng cả Mean hoặc Median.
- **Phân phối lệch phải (Right-skewed):** Đuôi kéo dài về phía giá trị lớn (vd: thu nhập, giá nhà); Mean > Median. Bắt buộc dùng Median để tránh bị kéo lệch bởi giá trị cực lớn.
- **Phân phối lệch trái (Left-skewed):** Đuôi kéo dài về phía giá trị nhỏ; Mean < Median. Nên dùng Median để đảm bảo tính đại diện.

**3. Kỹ thuật mã hóa biến (Encoding)**

- **One-hot encoding (Dummy encoding):** Tách cột định danh không có thứ tự (nominal) thành nhiều cột nhị phân (0 hoặc 1) tương ứng với từng giá trị thành phần.
- **Label / Ordinal Encoding:** Gán các số nguyên tương ứng cho biến có thứ tự hoặc cấp bậc rõ ràng.
- **Frequency / Target Encoding:** Thay thế nhãn bằng tần suất xuất hiện hoặc giá trị trung bình của biến mục tiêu đối với dữ liệu có quá nhiều nhãn khác nhau.

**Phân Tích Dữ Liệu Khám Phá (Exploratory Data Analysis - EDA)**

**1. Bản chất của EDA:**

- Là sự kết hợp giữa mô tả số học (description) và trực quan hóa (visualization) nhằm thấu hiểu phân phối, xu hướng và mối quan hệ trong dữ liệu.
- **Mục đích:** Khám phá insight, phát hiện vấn đề chất lượng dữ liệu. Không phải để chứng minh nhân quả (causation) và chưa bước sang giai đoạn huấn luyện mô hình máy học (ML).

**2. Bốn loại Phân tích Khám phá (Types of EDA):**

- **Univariate non-graphic:** Đơn biến phi đồ họa (phân tích 1 biến độc lập, tìm mẫu hình, không xem xét quan hệ).
- **Bivariate graphic:** Song biến có đồ họa (sử dụng biểu đồ thân-lá, histogram, box plot để nhìn bức tranh toàn diện từ 2 yếu tố).
- **Multivariate non-graphic:** Đa biến phi đồ họa (phân tích quan hệ từ 2 biến trở lên thông qua bảng chéo cross-tabulation hoặc thống kê).
- **Multivariate graphic:** Đa biến có đồ họa (sử dụng biểu đồ tán xạ scatterplot, biểu đồ bong bóng bubble chart, biểu đồ nhiệt heatmap, biểu đồ đa biến hoặc biểu đồ cột nhóm).

**3. Quy trình EDA chuẩn (6 bước):**

1. **Load:** Kiểm tra kích thước (shape) và ý nghĩa các cột.
2. **Missing values & invalid entries:** Xử lý triệt để giá trị thiếu và giá trị không hợp lệ.
3. **Outliers:** Phát hiện bằng IQR và quyết định xóa hay giới hạn (clip/winsorize).
4. **Distributions:** Mô tả thống kê (describe), vẽ histogram và countplot.
5. **Relationships:** Khám phá quan hệ qua heatmap, box/violin plot (lưu ý dùng tỷ lệ/rate thay vì số đếm tuyệt đối khi kích thước các nhóm khác nhau).
6. **Feature engineering:** Xử lý định dạng thời gian (datetime), tạo biến mới, phân nhóm (binning), mã hóa (dummies), chuẩn hóa (scaling).
7. **Summary:** Tóm tắt các insight chính rút ra kèm theo các điểm lưu ý hoặc giới hạn (caveats) của dữ liệu.

````python
Để học sâu phần này, Vũ cần hiểu **mỗi thao tác giải quyết vấn đề gì, áp dụng khi nào và có thể làm sai dữ liệu như thế nào**. Mình sẽ dùng các bảng dữ liệu nhỏ để giải thích theo mẫu:

**Tình huống → Nhận xét → Hướng xử lý → Kết luận.**

Các số liệu bên dưới là ví dụ giả lập.

**1. Define the question — xác định mục tiêu phân tích**

Giả sử có dữ liệu bán hàng:

| OrderID | Date | Branch | Product | Quantity | UnitPrice |
|---|---|---|---|---:|---:|
| O01 | 01/10/2026 | A | Coffee | 2 | 30.000 |
| O02 | 01/10/2026 | B | Tea | 1 | 25.000 |
| O03 | 02/10/2026 | A | Tea | 3 | 25.000 |

Trước khi đọc và xử lý dữ liệu, cần xác định:

| Câu hỏi | Ví dụ trả lời |
|---|---|
| Muốn biết gì? | Chi nhánh nào có doanh thu cao nhất trong tháng 10? |
| Kết quả phục vụ ai? | Quản lý kinh doanh |
| Đơn vị phân tích là gì? | Chi nhánh trong một tháng |
| Chỉ số nào trả lời câu hỏi? | Tổng doanh thu: `Quantity × UnitPrice` |
| Thành công được đo thế nào? | Tính được doanh thu từng chi nhánh, kiểm tra dữ liệu đủ và hợp lệ, chỉ ra chi nhánh dẫn đầu |

**Vì sao phải xác định trước?**

Cùng một bộ dữ liệu, mỗi mục tiêu cần cách xử lý khác nhau:

| Mục tiêu | Cần quan tâm |
|---|---|
| Tính tổng doanh thu | Không được tùy tiện loại đơn hàng lớn |
| Tìm mức chi tiêu điển hình | Median có thể phù hợp hơn mean |
| So sánh hiệu quả chi nhánh | Cần xét số ngày hoạt động, số đơn, quy mô |
| Dự đoán doanh thu | Cần chia dữ liệu train/test và đánh giá dự đoán |

Ví dụ chi nhánh A có doanh thu cao hơn B **chưa đủ để kết luận A hoạt động hiệu quả hơn** nếu A mở cửa nhiều ngày hơn hoặc có quy mô lớn hơn.

---

**2. Data loading — đọc dữ liệu và kiểm tra cấu trúc**

```python
import pandas as pd

df = pd.read_csv("sales.csv")
```

Excel:

```python
df = pd.read_excel("sales.xlsx", sheet_name="Orders")
```

Sau khi đọc, kiểm tra:

```python
print(df.shape)
print(df.head())
print(df.dtypes)
df.info()
```

| Lệnh | Giúp biết gì? |
|---|---|
| `shape` | Số dòng và số cột |
| `head()` | Một số bản ghi đầu |
| `dtypes` | Kiểu dữ liệu từng cột |
| `info()` | Cấu trúc, số giá trị không thiếu, kiểu dữ liệu |

**Tình huống: cột số đang là chuỗi**

| Quantity |
|---|
| `"2"` |
| `"3"` |
| `"unknown"` |

```python
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)
```

Kết quả:

| Quantity |
|---:|
| 2 |
| 3 |
| NaN |

**Nhận xét:** `errors="coerce"` chuyển giá trị không đọc được thành thiếu. Nó không sửa được số lượng thật.

**Hướng xử lý:** Kiểm tra các dòng bị chuyển thành thiếu rồi quyết định bổ sung, loại khỏi phép tính hoặc giữ để báo cáo.

**Tình huống: mã có số 0 ở đầu**

| StudentID |
|---|
| 00123 |
| 00124 |

Nếu đọc thành số, có thể mất số 0 đầu.

```python
df = pd.read_csv(
    "students.csv",
    dtype={"StudentID": "string"}
)
```

**Kết luận:** Mã định danh nên được xử lý theo ý nghĩa là mã, không phải biến số để tính toán.

**Kaggle Hub**

Quy trình cần hiểu là:

1. `dataset_download(...)` tải bộ dữ liệu.
2. Kết quả thường là đường dẫn thư mục.
3. Kiểm tra bên trong có file nào.
4. Dùng `read_csv`, `read_excel` hoặc hàm đọc phù hợp.

Không nên mặc định rằng tên dataset cũng là tên file CSV.

---

**3. `concat` và `merge` — hai kiểu ghép khác nhau**

**`concat`: nối các bảng cùng loại**

Bảng tháng 9:

| OrderID | Revenue |
|---|---:|
| O01 | 100 |
| O02 | 200 |

Bảng tháng 10:

| OrderID | Revenue |
|---|---:|
| O03 | 300 |
| O04 | 400 |

```python
sales = pd.concat(
    [sales_sep, sales_oct],
    ignore_index=True
)
```

Kết quả có bốn dòng.

**Tình huống cần kiểm tra:** Nếu cả hai file đều chứa O02, `concat` vẫn giữ cả hai bản ghi. Nó không tự xóa trùng.

**`merge`: ghép thông tin theo khóa**

Bảng đơn hàng:

| OrderID | CustomerID | Revenue |
|---|---|---:|
| O01 | C01 | 100 |
| O02 | C02 | 200 |
| O03 | C03 | 300 |

Bảng khách hàng:

| CustomerID | City |
|---|---|
| C01 | HCM |
| C02 | Hanoi |

```python
result = orders.merge(
    customers,
    on="CustomerID",
    how="left",
    validate="many_to_one",
    indicator=True
)
```

Kết quả:

| OrderID | CustomerID | Revenue | City |
|---|---|---:|---|
| O01 | C01 | 100 | HCM |
| O02 | C02 | 200 | Hanoi |
| O03 | C03 | 300 | NaN |

**Nhận xét:** City của C03 thiếu do không có khóa tương ứng ở bảng khách hàng.

**Hướng xử lý:** Kiểm tra mã C03 ở nguồn, không vội điền thành thành phố phổ biến nhất.

Pandas phân biệt `concat` để nối theo trục và `merge` để ghép theo khóa.

---

**4. Missingness mechanisms — cơ chế dữ liệu thiếu**

Cần phân biệt:

- **Cơ chế thiếu:** Vì sao giá trị không được quan sát?
- **Cách xử lý thiếu:** Điền, loại, bổ sung hay giữ nguyên?

Một cách điền phù hợp về kiểu dữ liệu chưa chắc phù hợp về cơ chế thiếu.

| Cơ chế | Hiểu bản chất | Tình huống minh họa |
|---|---|---|
| MCAR | Thiếu không phụ thuộc các giá trị dữ liệu liên quan | Hệ thống làm mất ngẫu nhiên một số câu trả lời |
| MAR | Sau khi xét các biến quan sát, việc thiếu không còn phụ thuộc giá trị đang thiếu | Trong từng nhóm năm học, thiếu thu nhập không phụ thuộc thu nhập; tỷ lệ thiếu khác nhau giữa các năm |
| MNAR | Việc thiếu vẫn phụ thuộc chính giá trị không quan sát được | Người có thu nhập rất cao thường không khai thu nhập |

**Ví dụ MNAR quan trọng**

Có 100 người:

- 80 người khai thu nhập.
- 20 người không khai vì thu nhập rất cao.

Nếu điền 20 ô thiếu bằng mean của 80 người:

> Thu nhập trung bình sau xử lý có thể bị đánh giá thấp, vì những người có thu nhập cao nhất chưa được quan sát.

**Hướng giải quyết:** Tìm hiểu quy trình thu thập, bổ sung thông tin nếu có thể, báo cáo giới hạn và xem kết quả thay đổi ra sao dưới các giả định khác nhau. Mean không tự giải quyết được MNAR.

**Có thể xác định MCAR/MAR/MNAR chỉ từ bảng không?**

Thường không thể khẳng định chỉ bằng việc nhìn bảng. Có thể tìm dấu hiệu:

```python
df["Income_missing"] = df["Income"].isna()

print(
    df.groupby("Year")["Income_missing"]
      .mean()
      .mul(100)
)
```

Nếu năm nhất thiếu nhiều hơn năm cuối:

> Việc thiếu thu nhập có liên quan đến năm học trong dữ liệu quan sát.

Điều đó chưa đủ chứng minh MAR.

---

**5. Chọn cách xử lý missing values**

Pandas dùng `isna()` để nhận diện, `fillna()` để điền và `dropna()` để loại dữ liệu thiếu.

Trước hết, lập bảng thiếu:

```python
missing = pd.DataFrame({
    "Count": df.isna().sum(),
    "Percent": df.isna().mean().mul(100)
})

print(missing)
```

**Trường hợp A: numeric thiếu ít, phân phối tương đối cân đối**

Dữ liệu:

```text
Age: 19, 20, NaN, 21, 20
```

Mean của các giá trị quan sát là 20:

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

**Kết luận:** Mean có thể phù hợp cho cách điền đơn giản nếu tuổi là biến đầu vào và việc thiếu không có dấu hiệu chọn lọc rõ ràng.

**Giới hạn:** Giá trị điền không phải tuổi thật; điền mean làm giảm độ phân tán.

**Trường hợp B: numeric lệch phải**

```text
Spending: 100, 110, 120, 130, 5000, NaN
```

| Chỉ số | Giá trị |
|---|---:|
| Mean | 1.092 |
| Median | 120 |

```python
df["Spending"] = df["Spending"].fillna(
    df["Spending"].median()
)
```

**Kết luận:** Median đại diện mức chi tiêu điển hình tốt hơn trong ví dụ này.

**Lưu ý:** Nếu đang tính doanh thu thực tế, không nên coi giá trị điền là giao dịch đã xác nhận.

**Trường hợp C: categorical thiếu ít**

```text
Transport: Bus, Bus, Motorbike, NaN, Bus
```

```python
modes = df["Transport"].mode()

if not modes.empty:
    df["Transport"] = df["Transport"].fillna(modes.iloc[0])
```

**Kết luận:** Có thể dùng mode cho bài tập điền đơn giản, nhưng sẽ làm tăng tỷ lệ nhóm phổ biến nhất.

**Trường hợp D: "Prefer not to say"**

Không nên tự suy đoán câu trả lời của người từ chối.

```python
df["Cafe"] = df["Cafe"].replace(
    "Prefer not to say",
    "Not disclosed"
)

df["Cafe"] = df["Cafe"].fillna("Unknown")
```

Giữ riêng hai nhóm nếu lý do thiếu có ý nghĩa:

| Giá trị | Ý nghĩa |
|---|---|
| `Not disclosed` | Chủ động không tiết lộ |
| `Unknown` | Chưa có thông tin |

Có thể gom chung `"Unknown"` nếu mục tiêu không cần phân biệt, nhưng phải ghi rõ quy ước.

**Trường hợp E: thiếu target**

```text
AtRisk: 0, 1, NaN, 0
```

```python
labeled = df.dropna(
    subset=["AtRisk"]
).reset_index(drop=True)
```

**Kết luận:** Không điền target bằng mode để tạo nhãn giả. Dòng thiếu nhãn vẫn có thể dùng cho phân tích không cần target.

**Trường hợp F: Age = −1**

`dropna()` không tự loại −1 vì −1 không phải giá trị thiếu.

```python
df.loc[df["Age"] < 0, "Age"] = float("nan")
```

Sau đó mới lựa chọn xử lý.

**Kết luận:** Giá trị sai ở một cột không nhất thiết khiến cả dòng vô dụng.

**Trường hợp G: thiếu rất nhiều**

Không có ngưỡng phổ quát "thiếu trên 50% phải xóa".

Nếu cột quan trọng nhưng thiếu 70%, việc xóa có thể khiến không trả lời được câu hỏi nghiên cứu. Cần cân nhắc bổ sung nguồn, phân tích riêng nhóm có dữ liệu hoặc thừa nhận chưa đủ thông tin.

**Phân biệt các lệnh:**

```python
df.dropna()                     # Loại dòng có bất kỳ ô thiếu
df.dropna(subset=["AtRisk"])   # Chỉ yêu cầu target không thiếu
df.dropna(how="all")           # Loại dòng thiếu toàn bộ
df.reset_index(drop=True)      # Đánh lại index
```

---

**6. Data wrangling — tổ chức dữ liệu thành dạng dùng được**

| Bước | Câu hỏi cần trả lời | Ví dụ |
|---|---|---|
| Discovery | Dữ liệu có gì, một dòng nghĩa là gì? | Một dòng là món hàng hay đơn hàng? |
| Structuring | Cấu trúc có phù hợp phân tích không? | Chuyển cột doanh thu từng tháng thành Month–Revenue |
| Cleaning | Có lỗi, trùng, sai kiểu không? | Sửa khoảng trắng, ngày sai |
| Enriching | Cần thêm thông tin nào? | Ghép tên khoa, tạo doanh thu |
| Validating | Dữ liệu sau xử lý có đúng không? | Kiểm tra khóa và miền giá trị |
| Publishing | Người khác sử dụng được không? | Xuất bảng sạch và ghi chú xử lý |

**Ví dụ structuring:**

Dạng ban đầu:

| Branch | Jan | Feb |
|---|---:|---:|
| A | 100 | 150 |
| B | 200 | 220 |

Chuyển sang:

| Branch | Month | Revenue |
|---|---|---:|
| A | Jan | 100 |
| B | Jan | 200 |
| A | Feb | 150 |
| B | Feb | 220 |

```python
long_df = df.melt(
    id_vars="Branch",
    var_name="Month",
    value_name="Revenue"
)
```

**Kết luận:** Đây là thay đổi cấu trúc, không phải sửa lỗi. Dạng dài thuận tiện để nhóm và vẽ biểu đồ theo tháng.

---

**7. Outliers — phát hiện trước, quyết định sau**

\[
IQR=Q_3-Q_1
\]

\[
L=Q_1-1.5IQR,\qquad U=Q_3+1.5IQR
\]

```python
q1 = df["Revenue"].quantile(0.25)
q3 = df["Revenue"].quantile(0.75)

iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr
```

Giá trị ngoài hai ngưỡng là **ngoại lệ theo quy tắc IQR**, chưa chắc là lỗi.

| Tình huống | Hướng xử lý |
|---|---|
| Tuổi −1 | Kiểm tra, chuyển thành thiếu nếu không sửa được |
| Một đơn hàng cực lớn nhưng có hóa đơn | Giữ khi tính doanh thu |
| Ngủ 25 giờ/ngày | Kiểm tra lỗi nhập hoặc đơn vị |
| Lương cao của quản lý cấp cao | Có thể hợp lệ; dùng median hoặc phân tích theo nhóm |
| Cân nặng lệch 1.000 lần | Kiểm tra đơn vị trước khi tự sửa |

**Drop:** loại bản ghi.

```python
filtered = df.loc[
    df["Revenue"].between(lower, upper)
].copy()
```

**Clip:** giữ dòng, giới hạn giá trị.

```python
df["Revenue_clipped"] = df["Revenue"].clip(lower, upper)
```

**Winsorize:** thường giới hạn hai đầu theo phân vị.

```python
low = df["Revenue"].quantile(0.01)
high = df["Revenue"].quantile(0.99)

df["Revenue_winsorized"] = df["Revenue"].clip(low, high)
```

**Kết luận:** Không tự động xóa ngoại lệ. Phải xác định đó là lỗi hay hiện tượng thật và xem mục tiêu phân tích có cần giữ nó không.

---

**8. EDA — xem phân phối và quan hệ**

Tài liệu ghi "6-step EDA" nhưng thực tế liệt kê **7 mục**, có thể hiểu là sáu bước phân tích và một bước tổng kết.

**Phân phối một biến**

```python
df["Revenue"].describe()
```

| Chỉ số | Ý nghĩa |
|---|---|
| `count` | Số giá trị không thiếu |
| `mean` | Trung bình |
| `std` | Độ lệch chuẩn |
| `min`, `max` | Giá trị nhỏ nhất, lớn nhất |
| `25%`, `50%`, `75%` | Các phân vị; `50%` là median |

**Tình huống:** Mean cao hơn median nhiều.

**Nhận xét:** Có dấu hiệu lệch phải, có thể do một số giá trị lớn.

**Hướng xử lý:** Xem histogram, kiểm tra giá trị lớn; cân nhắc median khi mô tả mức điển hình.

**So sánh quan hệ: dùng tỷ lệ khi nhóm khác quy mô**

| Khoa | Tổng sinh viên | Số trượt | Tỷ lệ trượt |
|---|---:|---:|---:|
| A | 1.000 | 100 | 10% |
| B | 100 | 20 | 20% |

**Kết luận:**

- A có nhiều sinh viên trượt hơn về số lượng.
- B có tỷ lệ trượt cao hơn.

```python
summary = df.groupby("Faculty").agg(
    Total=("Failed", "count"),
    FailedCount=("Failed", "sum"),
    FailedRate=("Failed", "mean")
)
```

Điều kiện: `Failed` đã được kiểm tra là 0/1. Nếu có nhãn thiếu, tỷ lệ này chỉ tính trên người có nhãn.

**Correlation không phải causation**

Ví dụ giờ học và điểm có tương quan dương:

> Trong mẫu, người học nhiều giờ có xu hướng có điểm cao hơn.

Chưa đủ để nói:

> Tăng giờ học chắc chắn làm điểm tăng.

Có thể còn năng lực nền, phương pháp học, độ khó môn học và sai lệch chọn mẫu.

Tương quan gần 0 cũng không chứng minh không có quan hệ; quan hệ có thể phi tuyến.

---

**9. Feature engineering — tạo biến có ý nghĩa**

| Kỹ thuật | Ví dụ | Mục đích |
|---|---|---|
| Datetime | Tách tháng từ ngày mua | Phân tích theo thời gian |
| New variables | `Revenue = Quantity × UnitPrice` | Tạo chỉ số nghiệp vụ |
| Binning | Chia tuổi thành nhóm | So sánh nhóm |
| Dummies | Chuyển phương tiện thành cột 0/1 | Chuẩn bị đầu vào cho mô hình |
| Scaling | Chuẩn hóa biến số | Phục vụ một số thuật toán |

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%d/%m/%Y",
    errors="coerce"
)

df["Month"] = df["Date"].dt.to_period("M")

df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

**Binning có đánh đổi:** Dễ diễn giải nhưng mất chi tiết. Hai người 18 và 24 tuổi có thể bị gom cùng nhóm dù khác nhau đáng kể.

**Dummies và scaling không bắt buộc cho EDA.** Nếu dùng cho ML, các tham số điền thiếu và scaling phải học từ tập train rồi áp dụng sang test.

---

**10. Visualization — chọn biểu đồ theo câu hỏi**

| Muốn xem gì? | Biểu đồ |
|---|---|
| Phân phối một biến số | Histogram |
| Số lượng từng nhóm | Countplot |
| Thống kê theo nhóm | Barplot |
| Xu hướng theo thời gian | Line |
| Quan hệ hai biến số | Scatter |
| Phân phối biến số giữa các nhóm | Boxplot / violin |
| Ma trận tương quan | Heatmap |
| Quan hệ nhiều biến số | Pairplot |

**Countplot và barplot khác nhau:**

```python
sns.countplot(data=df, x="Branch")
```

Đếm số bản ghi từng chi nhánh.

```python
sns.barplot(data=df, x="Branch", y="Revenue")
```

Mặc định thể hiện **doanh thu trung bình** từng chi nhánh, không phải tổng doanh thu.

Nếu cần tổng doanh thu, tính trước:

```python
revenue_by_branch = (
    df.groupby("Branch", as_index=False)["Revenue"].sum()
)

sns.barplot(
    data=revenue_by_branch,
    x="Branch",
    y="Revenue",
    errorbar=None
)
```

**Matplotlib và Seaborn phối hợp:**

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(
    data=df,
    x="Revenue",
    ax=axes[0]
)

sns.boxplot(
    data=df,
    x="Branch",
    y="Revenue",
    ax=axes[1]
)

axes[0].set_title("Phân phối doanh thu")
axes[1].set_title("Doanh thu mỗi bản ghi theo chi nhánh")

plt.tight_layout()
plt.show()
```

`hue` thêm phân nhóm bằng màu; `col` và `row` tạo các ô biểu đồ riêng để so sánh.

---

**11. Cách viết kết luận và hướng giải quyết**

Một kết luận tốt phải phân biệt **phát hiện trong dữ liệu** với **đề xuất hành động**.

Ví dụ:

> Chi nhánh B có doanh thu trung bình mỗi đơn thấp hơn A, nhưng số đơn nhiều hơn nên tổng doanh thu cao hơn. Cần kiểm tra cơ cấu sản phẩm và thời gian hoạt động trước khi đánh giá hiệu quả. Có thể phân tích thêm doanh thu mỗi ngày và giá trị đơn theo loại sản phẩm.

Mẫu trả lời dùng trong bài tập:

| Ý | Nội dung cần viết |
|---|---|
| Vấn đề | Cột nào thiếu, sai hoặc bất thường? |
| Ảnh hưởng | Có thể làm lệch kết quả nào? |
| Xử lý | Chọn cách nào và vì sao? |
| Kiểm tra | Sau xử lý có đúng miền, đúng số dòng, đúng khóa không? |
| Kết luận | Dữ liệu cho thấy gì, còn giới hạn gì? |

Ví dụ hoàn chỉnh:

> Cột thu nhập có 8% giá trị thiếu và phân phối lệch phải. Điền mean có thể tạo giá trị cao hơn mức thu nhập điển hình. Nếu mục tiêu là chuẩn bị biến đầu vào và chưa thấy dấu hiệu thiếu có tính chọn lọc, có thể điền median, đồng thời đánh dấu các ô được điền. Sau xử lý cần kiểm tra lượng thiếu và so sánh phân phối trước–sau. Không coi giá trị điền là thu nhập thực tế đã xác nhận.

**Điểm cần nhớ nhất:** Không có thao tác làm sạch luôn đúng. Cách xử lý phải dựa vào **mục tiêu phân tích, ý nghĩa biến, nguyên nhân thiếu và ảnh hưởng đến kết luận**.
````
